# Plotting Results

Generates and saves all figures from CSV results files.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from src.partition import load_partition, get_label_distribution
from src.utils import load_mnist
from src.plots import (
    plot_label_distribution,
    plot_centralized_curves,
    plot_fedsgd_participation,
    plot_fedavg_computation,
    plot_noniid_comparison,
    plot_interpolation,
    plot_survey_curves
)

os.makedirs('plots', exist_ok=True)
(train_x, train_y), (test_x, test_y) = load_mnist('data')
```

Plot partition label distributions.

```python
iid_p = load_partition('data/iid_k100.npz')
plot_label_distribution(get_label_distribution(train_y, iid_p), 'IID Partition Label Distribution', 'plots/iid_dist.png')

patho_p = load_partition('data/pathological_k100.npz')
plot_label_distribution(get_label_distribution(train_y, patho_p), 'Pathological Partition Label Distribution', 'plots/pathological_dist.png')

dir_p = load_partition('data/dirichlet_a05.npz')
plot_label_distribution(get_label_distribution(train_y, dir_p), 'Dirichlet (alpha=0.5) Label Distribution', 'plots/dirichlet_dist.png')
print('Partition plots saved.')
```

Plot centralized baseline curves.

```python
plot_centralized_curves('results/centralized.csv', 'plots/centralized.png')
print('Centralized plot saved.')
```

Plot FedSGD participation fraction and FedAvg computation effect.

```python
plot_fedsgd_participation('results/fedsgd_participation.csv', 'plots/fedsgd_participation.png')
plot_fedavg_computation('results/fedavg_computation.csv', 'plots/fedavg_computation.png')
print('FedSGD and FedAvg plots saved.')
```

Plot FedAvg on non-IID partitions.

```python
plot_noniid_comparison('results/noniid_experiments.csv', 'plots/noniid.png')
print('Non-IID comparison plot saved.')
```

Plot common vs independent initialisation interpolation.

```python
plot_interpolation('results/common_init.csv', 'plots/common_init.png')
print('Common init plot saved.')
```

Plot survey demos: settings, non-IID types, dropout, communication cost, and client drift.

```python
plot_survey_curves('results/survey_settings.csv', 'plots/settings.png',
                   group_col='setting', title='Cross-Silo vs Cross-Device', y_label='Test accuracy (%)')

plot_survey_curves('results/survey_noniid_types.csv', 'plots/noniid_types.png',
                   group_col='skew_type', title='Non-IID Heterogeneity Types', y_label='Test accuracy (%)')

plot_survey_curves('results/survey_dropout.csv', 'plots/partial_participation.png',
                   group_col='drop_prob', title='Effect of Client Dropout', y_label='Test accuracy (%)')

plot_survey_curves('results/survey_communication.csv', 'plots/communication_cost.png',
                   x_col='megabytes', y_col='test_acc', group_col='setting',
                   title='Accuracy vs Communication Volume', x_label='Total MB transferred', y_label='Test accuracy (%)')

plot_survey_curves('results/survey_client_drift.csv', 'plots/client_drift.png',
                   x_col='round', y_col='cosine_similarity', group_col='partition',
                   title='Client Update Cosine Similarity (Client Drift)', x_label='Round', y_label='Mean cosine similarity')

print('All survey plots generated successfully.')
```
